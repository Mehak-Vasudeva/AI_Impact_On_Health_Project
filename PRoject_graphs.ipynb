```python
df= pd.read_csv('AI_Impact_on_Health_Dataset_5000_Enhanced.csv')
```

```python
columns= ['Age_Group','Gender','BMI_Category','AI_Tool_Type','Exercise_Level','Satisfaction_Level','Risk_Category','Improvement_Category']
for col in columns:
    df[col]=df[col].astype('category')
```

```python
df['AI_Usage_Level']= df['AI_Usage_Level'].astype('category')
```

```python
df['Chronic_Condition']=df['Chronic_Condition'].map({'Yes': True, 'No': False})
```

```python
import plotly.express as px
```

```python
fig=px.histogram(df,
            x='AI_Tool_Type',
            color='Gender',
            nbins=20,
            title="AI tool Usage by Gender",
            text_auto=True,
            # color_discrete_sequence=["#2b8cbe", "#e34a33", "#636363"],
            color_discrete_sequence=["#aec7e8","#ffbb78","#98df8a"],
            # opacity = 0.5

            
)
fig.update_layout(
    title_x=0.5,
    xaxis_title='AI Tool Type',
    yaxis_title='Count',
    bargap=0.05,
    font=dict(size=14)
)
fig.update_traces(
    marker_line_color="black",
    marker_line_width= 1

)
fig.show()
```

```python
fig= px.box(
    df,
    x="Total_Engagement_Hours_Per_Week",
    y="Improvement_Percentage",
    labels={
    "Total_Engagement_Hours_Per_Week":"Engagement Hours Per Week",
    "Improvement_Percentage":"Improvement Percentage",
    },
    

    title="Engagement Hours vs Improvement Percentage"
)
fig.update_layout(
    title_x=0.5,
    font=dict(size=14),
)
fig.update_traces(
    marker_size=5
)
fig.show()
```

```python
dark_palette = ["#2b5c8f", "#4682b4", "#76a5af", "#e06666"]
fig=px.pie(df,names="Improvement_Category",title="Proportion of Improved Categories",hole=0.4,color_discrete_sequence=dark_palette
)
fig.update_traces(textinfo="value + percent")
```

```python
df_ = df.groupby('AI_Tool_Type')['Improvement_Percentage'].mean().reset_index()

fig1 = px.bar(
    df_, 
    x='AI_Tool_Type', 
    y='Improvement_Percentage', 
    color='AI_Tool_Type',
    color_discrete_sequence=["#5B8FF9","#5AD8A6","#5D7092","#F6BD16"],
    title='Average Health Improvement (%) by AI Tool Type',
    labels={'Improvement_Percentage': 'Avg Improvement (%)', 'AI_Tool_Type': 'AI Tool Type'},
    template='plotly_white'
)
fig1.update_layout(showlegend=False)
fig1.show()
```

```python
fig2 = px.pie(
    df, 
    names='AI_Tool_Type', 
    hole=0.45,
    title='Distribution of AI Tool Usage',
    color_discrete_sequence=px.colors.qualitative.T10, 
    template='plotly_white'
)
fig2.update_traces(textposition='inside', textinfo='percent+label')
fig2.show()
```

```python
fig3 = px.histogram(
    df, 
    x='Age', 
    color='Gender', 
    barmode='group',
    nbins=30,
    color_discrete_map={'Male': '#4E79A7', 'Female': '#F28E2B', 'Other': '#E15759'},
    title='Demographic Distribution: Age Breakdown by Gender',
    template='plotly_white'
)
fig3.update_traces(opacity=0.75)
fig3.show()
```

```python
fig = px.box(
    df, 
    x='BMI_Category', 
    y='Improvement', 
    color='BMI_Category',
    color_discrete_sequence=["#2b5c8f","#d95f02","#7570b3","#e7298a"],
    title='Health Score Improvement Distribution across BMI Categories',
    labels={'Improvement': 'Absolute Health Score Improvement', 'BMI_Category': 'BMI Category'},
    template='plotly_white'
)
fig.show()
```

```python
fig = px.violin(
    df, 
    x='AI_Usage_Level', 
    y='Satisfaction_Score', 
    color='AI_Usage_Level', 
    box=True, 
    points='outliers',
    color_discrete_sequence=px.colors.qualitative.Prism,
    title='Satisfaction Score Distribution across AI Usage Levels',
    labels={'Satisfaction_Score': 'Satisfaction Score', 'AI_Usage_Level': 'AI Usage Level'},
    template='plotly_white'
)
fig.show()
```

```python

df_ = df.groupby('Age_Group')[['Health_Score_Before_AI', 'Health_Score_After_AI']].mean().reset_index()

fig = px.line(
    df_, 
    x='Age_Group', 
    y=['Health_Score_Before_AI', 'Health_Score_After_AI'],
    markers=True,
    color_discrete_sequence=["#E15759", '#4E79A7'],
    title='Average Health Score Before vs. After AI by Age Group',
    labels={'value': 'Mean Health Score', 'variable': 'Stage', 'Age_Group': 'Age Group'},
    template='plotly_dark'
)

fig.show()
```

```python
corr_cols = [
    'Age', 'BMI', 'AI_Health_App_Usage_Hours_Per_Week', 
    'Exercise_Hours_Per_Week', 'Health_Score_Before_AI', 
    'Health_Score_After_AI', 'Improvement_Percentage'
]
corr_matrix = df[corr_cols].corr()

fig = px.imshow(
    corr_matrix, 
    text_auto='.2f', 
    color_continuous_scale='RdBu',
    # color_continuous_midpoint=0.2,
    title='Correlation Heatmap',
    template='plotly_white',
        
)
fig.update_layout(
    width=900,
    height=800,
    
    title={
        'text': "Correlation Heatmap",
        'x': 0.5,                # 0.5 ka matlab hai bilkul centre mein
        'xanchor': 'center',     # Title ko uske centre point se align karega
        'font': {
            'size': 26,          # Font size ko bada karne ke liye (Default 17 hota hai)
            'color': 'black'     # Agar dark theme use kar rahe hain toh white rakhein
        }
    },

)
fig.show()
```

```python
fig = px.treemap(
    df, 
    path=['BMI_Category', 'AI_Tool_Type'], 
    values='Improvement',
    color='Improvement',
    color_continuous_scale='RdYlGn',
    title='Treemap: Health Improvement Volume by BMI Category and AI Tool',
    template='plotly_white'
)
fig.show()
```
